## Setup — generate `week3_students.csv` (stand-in dataset)

```python
import random
random.seed(42)

first_names = ["Aline","Eric","Diane","Jean Bosco","Sandrine","Patrick","Claudine","Emmanuel",
                "Christelle","Olivier","Alice","Fabrice","Grace","Innocent","Josiane","Kevin",
                "Liliane","Moise","Nadine","Oscar","Peace","Queen","Robert","Sarah","Theogene",
                "Umutoni","Vanessa","Willy","Xavera","Yves","Zeina","Aimee","Bruce","Chantal",
                "Damien","Esperance","Felix","Gisele","Hakizimana","Immaculee"]
last_names = ["Uwase","Niyonzima","Mukamana","Habimana","Ingabire","Mugisha","Uwimana","Nkurunziza",
              "Mutesi","Ndayisenga","Umuhoza","Bizimana","Nkusi","Iradukunda","Mahoro","Byiringiro",
              "Nyirahabimana","Rugamba","Kayitesi","Twagirayezu"]
programs = ["Information Systems","Software Engineering","Networking","Accounting"]
districts = ["Gasabo","Musanze","Muhanga","Kicukiro","Huye","Nyagatare","Rubavu","Nyanza"]

rows = []
for i in range(40):
    sid = f"AUCA{i+1:03d}"
    name = f"{first_names[i]} {last_names[i % len(last_names)]}"
    age = random.randint(18, 27)
    gender = random.choice(["F", "M"])
    program = programs[i % len(programs)]
    district = districts[i // 5]          # 5 students per district, in order
    attendance = random.randint(60, 99)
    score = random.randint(35, 98)
    rows.append([sid, name, age, gender, program, district, attendance, score])


broken_indices = [7, 19, 33]
broken_values = ["N/A", "absent", "--"]
for idx, val in zip(broken_indices, broken_values):
    rows[idx][7] = val

header = "student_id,name,age,gender,program,district,attendance_pct,score"
with open("week3_students.csv", "w") as f:
    f.write(header + "\n")
    for r in rows:
        f.write(",".join(str(x) for x in r) + "\n")

print("week3_students.csv created with", len(rows), "rows.")
```

### Exercise 1.1

```python
def print_grade_ladder(score):
    if score >= 85:
        grade = "A"
    elif score >= 70:
        grade = "B"
    elif score >= 60:
        grade = "C"
    elif score >= 50:
        grade = "D"
    else:
        grade = "F"
    print(f"Score {score} -> Grade {grade}")

for s in [91, 76, 64, 52, 43]:
    print_grade_ladder(s)
```

def validate(score, attendance):
    if score >= 50 and attendance >= 75:
        result = "Validated"
    elif score >= 50 and attendance < 75:
        result = "Blocked: low attendance"
    else:
        result = "Failed"
    print(f"score={score}, attendance={attendance} -> {result}")

for score, att in [(76, 92), (76, 68), (43, 95)]:
    validate(score, att)

## Part 3 · Functions
### Exercise 3.1 — get_grade

```python
def get_grade(score):
    """Return the letter grade (A/B/C/D/F) for a numeric score."""
    if score >= 85:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    elif score >= 50:
        return "D"
    else:
        return "F"

for s in [91, 76, 43]:
    print(s, "->", get_grade(s))
```

### Exercise 3.2 


```python
def pass_rate(score_list):
    """Return the percentage of scores >= 50 in score_list."""
    passed = 0
    count = 0
    for s in score_list:
        count += 1
        if s >= 50:
            passed += 1
    return (passed / count) * 100

print("Pass rate:", pass_rate(scores))
```

## Part 5 · 
### Exercise 5.1

```python
valid = 0
bad = 0
bad_ids = []
total_valid_score = 0

for line in lines[1:]:
    parts = line.strip().split(",")
    student_id = parts[0]
    try:
        score = int(parts[7])
        valid += 1
        total_valid_score += score
    except ValueError:
        bad += 1
        bad_ids.append(student_id)

average_valid = total_valid_score / valid

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average of valid scores:", round(average_valid, 1))
```

### Exercise 5.2 — Grades & the top student

```python
passed_count = 0
failed_count = 0
best_score = -1
best_name = ""

for line in lines[1:]:
    parts = line.strip().split(",")
    try:
        score = int(parts[7])
    except ValueError:
        continue

    if score >= 50:
        passed_count += 1
    else:
        failed_count += 1

    if score > best_score:
        best_score = score
        best_name = parts[1]

print("Passed:", passed_count)
print("Failed:", failed_count)
print("Top student:", best_name, "with score", best_score, "-> grade", get_grade(best_score))
```

### Exercise 5.3 — Stretch: report by district

```python
sums = {}
counts = {}

for line in lines[1:]:
    parts = line.strip().split(",")
    d = parts[5]
    try:
        s = int(parts[7])
    except ValueError:
        continue

    sums[d] = sums.get(d, 0) + s
    counts[d] = counts.get(d, 0) + 1

best_district = ""
best_avg = -1

for d in sums:
    avg = sums[d] / counts[d]
    print(f"{d}: average = {round(avg, 1)}")
    if avg > best_avg:
        best_avg = avg
        best_district = d

print(f"\nHighest average district: {best_district} ({round(best_avg, 1)})")
```
